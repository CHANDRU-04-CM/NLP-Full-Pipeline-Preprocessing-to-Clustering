```python
# Import Libraries
import nltk
import string
import numpy as np

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.cluster import KMeans
```

```python
# Download required NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
text = """
Artificial Intelligence (AI) is transforming the world! It is used in healthcare, finance, education, and transportation.
Machines can learn from data, analyze patterns, and make decisions. However, AI also raises questions: Will machines replace humans?

Natural Language Processing (NLP) is a branch of AI. It helps computers understand human language; for example, chatbots, translation systems, and voice assistants.
NLP techniques include tokenization, stemming, lemmatization, and stop-word removal. These techniques improve text analysis!
"""
```

```python
print("\nOriginal Text:\n")
print(text)
```

```python
# Remove Punctuation

text_no_punct = text.translate(str.maketrans('', '', string.punctuation))

print("\nText After Removing Punctuation:\n")
print(text_no_punct)
```

```python
# Tokenization

tokens = word_tokenize(text_no_punct)

print("\nTokens:\n")
print(tokens)
```

```python
#Stop Word Removal

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in tokens if word.lower() not in stop_words]

print("\nAfter Stop Word Removal:\n")
print(filtered_words)
```

```python
#Stemming

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\nAfter Stemming:\n")
print(stemmed_words)
```

```python
# Lemmatization

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nAfter Lemmatization:\n")
print(lemmatized_words)
```

```python
#Bag of Words

cv = CountVectorizer()
bow = cv.fit_transform([text])

print("Feature Names:\n", cv.get_feature_names_out())
print("\nBoW Matrix:\n", bow.toarray())
```

```python
#TF-IDF

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform([text])

print("Feature Names:\n", tfidf.get_feature_names_out())
print("\nTF-IDF Matrix:\n", tfidf_matrix.toarray())
```

```python
# TEXT CLASSIFICATION (SUPERVISED)

docs = [
    "AI is powerful",
    "Machine learning is useful",
    "I love programming",
    "Python is great",
    "I hate bugs",
    "Errors are frustrating"
]
```

```python
 # 1=Positive
 # 0=Negative
labels = [1, 1, 1, 1, 0, 0]
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(docs)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.2, random_state=42)
```

```python
model = MultinomialNB()
model.fit(X_train, y_train)
```

```python
print("\n CLASSIFICATION PREDICTION ")
print(model.predict(X_test))
```
